```python
"""03-01 실습 정답 예시: 목표별 신호와 태그 보유 상태를 대조합니다."""

from pathlib import Path

import pandas as pd



df = pd.read_csv("..//3일차_실습/03-01_회전기계_태그목록_실습.csv").set_index("tag")

cases = {
    "CASE A - 모터 베어링 손상": [
        ("진동 가속도 PEAK", "MTR01_VIB_ACC", "주 신호: 초기 베어링 충격에 민감"),
        ("수평 진동 속도 RMS", "MTR01_VIB_H", "방사 방향 진동의 확인 신호"),
        ("전류", "MTR01_CURRENT", "부하·마찰 변화 확인"),
        ("권선 온도", "MTR01_TEMP", "진행된 열화의 열 반응 확인"),
        ("회전수", "MTR01_RPM", "운전 조건 통제"),
    ],
    "CASE B - 팬 분진 축적 불평형": [
        ("수평 진동 속도 RMS", "FAN01_VIB_H", "주 신호: 느린 불평형 추세"),
        ("전류", "FAN01_CURRENT", "부하 반응 확인"),
        ("베어링 온도", "FAN01_TEMP", "보조 상태 신호"),
        ("회전수", "FAN01_RPM", "운전 속도 영향 분리; 필수 요청"),
        ("수직 진동", "FAN01_VIB_V", "방향 비교; 권장 요청"),
        ("축방향 진동", "FAN01_VIB_A", "정렬 등 대안 원인 구별; 권장 요청"),
    ],
}

for title, rows in cases.items():
    print(f"\n=== {title} ===")
    result = []
    for requirement, tag, reason in rows:
        available = tag in df.index and df.loc[tag, "보유여부"] == "O"
        result.append(
            {
                "필요 신호": requirement,
                "태그": tag,
                "보유": "O" if available else "X",
                "판단 근거": reason,
            }
        )
    print(pd.DataFrame(result).to_string(index=False))

print(
    "\n모터 베어링 초기 손상은 가속도 PEAK 또는 원시 파형/주파수 성분이 더 적합할 수 있습니다. "
    "현재 요약값만으로 조기 손상을 확정할 수는 없습니다."
)
print(
    "팬의 장기 진동 감시는 가능하지만 회전수 태그가 없어 운전 속도 변화와 불평형 영향을 "
    "분리하기 어렵습니다. 태그 목록은 실습을 위한 예시 인벤토리입니다."
)
```
